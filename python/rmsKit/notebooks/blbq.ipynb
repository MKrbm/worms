```python
import sys
import os
sys.path.insert(0, "../")
import torch
from lattice import BLBQ
# # os.environ["CUDA_VISIBLE_DEVICES"] = ""
# # os.environ["XLA_PYTHON_CLIENT_PREALLOCATE"] = "false"
import rms_torch
import numpy as np
```

```python
p = dict(
    J0 = 1,
    J1 = -0.1,
    hx=0.5,
    hz=0,
    lt = -1,
    obc = True,
)


H = BLBQ.system([6], p)[0]
h_loc = BLBQ.local(p)

```

```python
off = np.ones(H.shape[0]) * 100
off = np.diag(off)
H_abs = H - off
H_abs = -np.abs(H_abs)
H_abs = H_abs + off
E_abs, V_abs = np.linalg.eigh(H_abs)
E, V = np.linalg.eigh(H)
```

```python

T = 0.25
z = np.exp(-E / T)
# Calculate average energy
e_exp = (E * z).sum() / z.sum()
# Calculate average energy squared
e_squared_exp = (E**2 * z).sum() / z.sum()
# Calculate specific heat (C_v = (⟨E²⟩ - ⟨E⟩²)/T²)
c_v = (e_squared_exp - e_exp**2) / (T**2)

z_abs = np.exp(-E_abs / T)

AS = z.sum()/z_abs.sum()

print(AS, e_exp / 4, c_v / 4)
```
