```python
import rms_torch
import torch
import geoopt
importlib.reload(rms_torch)
```

```python
sps = 2
h_size = sps ** 2
device = torch.device("cpu")
model = rms_torch.UnitaryRiemann(
    h_size, sps, device=device,
    manifold=geoopt.CanonicalStiefel(),
    ).to(device)



# generate random hamiltonian

H = torch.randn(h_size, h_size, device=device, dtype=torch.float64)
H = H + H.T

loss = rms_torch.MinimumEnergyLoss([H], device=device)
```

```python
optimizer = rms_torch.optimizer.RiemannianAdam(params=[model.u], lr=0.01, betas=(0.9, 0.999))
```

```python
for _ in range(1000):
    optimizer.zero_grad()
    U = model()

    HU = U @ H @ U.T
    loss_val = loss(HU)

    loss_val.backward()

    optimizer.step()

    print(loss_val)

```
